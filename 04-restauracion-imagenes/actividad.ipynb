Para la primera imagen se solicita la aplicación de ruido gaussiano. Luego de realizar una copia de la imagen original recortada, se normaliza al rango [0, 1] para trabajar con valores flotantes que permitan la suma del ruido sin salirse del rango válido. Posteriormente, se genera una matriz de ruido con distribución normal, donde el parámetro "loc" indica la media del ruido (en este caso 0.5), y "scale" representa la desviación estándar, que define qué tan dispersos están los valores respecto a la media. Finalmente, se suma este ruido a la imagen normalizada, lo que provoca que cada píxel se vea alterado de forma aleatoria, pero siguiendo una distribución en forma de campana.

###Filtro Media Geometrica (filtro lineal)

Para la primera imagen con ruido gaussiano se utilizó un filtro lineal llamado media geométrica. Este código fue extraído de la documentación disponible en el Google Colab de la clase. Se eligió este filtro porque el ruido gaussiano afecta los valores de los píxeles de forma simétrica en toda la imagen. La media geométrica suaviza eficazmente la imagen sin distorsionar los bordes, preservando el contenido original mientras reduce el ruido fino, que es el principal causante del oscurecimiento visual en la imagen.
Este filtro reemplaza el valor central de la mascara por el promedio de los valores contenidos en ella, en este caso 3x3.

```python
import cv2
import numpy as np
import scipy.ndimage as ndi
import matplotlib.pyplot as plt

def filtro_geometrico(A):
    largo=  len(A)
    S= np.prod(A.flatten())**(1/largo)
    return S

filtro1= ndi.generic_filter(res1,filtro_geometrico, [3,3])

fig, ax = plt.subplots(1,2)
fig.set_size_inches(15,15)
ax[0].imshow(res1, cmap='gray')
ax[0].set_title(f"img con ruido Gaussiano")
ax[1].imshow(filtro1, cmap='gray')
ax[1].set_title(f"img filtrado geometrico")
plt.show()
```

###filtro promedio alfa-acotado (filtro de orden estadistico)

Para la imagen con ruido uniforme se utilizó un filtro de orden estadístico llamado promedio alfa-acotado. Al igual que el anterior, este código fue extraído del Google Colab de la asignatura. Se escogió este filtro porque el ruido uniforme introduce valores aleatorios dentro de un rango fijo, y este método elimina los valores más altos y más bajos dentro de cada ventana local, que suelen corresponder al ruido. Al conservar solo los valores intermedios, permite suavizar la imagen sin perder tanto detalle.

```python
import cv2
import numpy as np
import scipy.ndimage as ndi
import matplotlib.pyplot as plt


def filtro_promedio_podado(subimage, acut):
    S= np.sort(subimage.flatten())
    B= S[acut:-acut]
    val=np.sum(B)/len(B)
    return val


#aplicamos el filtro promedio alfa-acotado
filtro2= ndi.generic_filter(res2,filtro_promedio_podado, [4,4], extra_keywords={'acut':2})

fig, ax = plt.subplots(1,2)
fig.set_size_inches(15,15)
ax[0].imshow(res2, cmap='gray')
ax[0].set_title(f"img con ruido Uniforme")
ax[1].imshow(filtro2, cmap='gray')
ax[1].set_title(f"img con filtro alfa-acotado")
plt.show()
```

###Filtro de mediana adaptativa (filtro adaptativo)

Para la tercera imagen, que presenta ruido impulsional tipo Salt, se utilizó un filtro adaptativo llamado filtro de mediana adaptativa. Se eligió este filtro porque el ruido “salt” aparece como píxeles blancos extremos (valor 255) distribuidos aleatoriamente en la imagen. El filtro detecta estos valores atípicos y los reemplaza sin afectar los demás píxeles que no están contaminados por el ruido. Funciona recorriendo la imagen con una ventana que se adapta dinámicamente: si el tamaño inicial no permite una decisión fiable, se incrementa hasta encontrar un vecindario apropiado. Esta adaptabilidad lo hace especialmente eficaz en imágenes con alta densidad de ruido impulsivo, ya que preserva mejor los bordes y detalles que un filtro de mediana fijo.

```python
import cv2
import numpy as np
import scipy.ndimage as ndi
import matplotlib.pyplot as plt

def filtro_mediana_adaptiva(A):
    m,n = A.shape
    Smax = 16
    ws   = 4
    J = np.zeros([m,n])

    for i in range(0,m-Smax):
        for j in range(0,n-Smax):
            sw     = 1
            Zxy    = A[i,j]
            wadapt = ws
            while(sw):
                B = A[i:i+wadapt, j:j+wadapt]

                pixel, newwin = bloque(B,Zxy,wadapt,Smax)
                if ( pixel != -1):
                    J[i,j]=pixel
                    sw = 0
                else:
                    wadapt=newwin

    return (np.uint8(J[0:m-Smax, 0:n-Smax]))
def bloque(A,Zxy, wadapt, Smax):

    newwin= wadapt

    Zmin= np.min(A.flatten())
    Zmax= np.max(A.flatten())
    Zmed= np.median(A.flatten())
    px = -1

    A1= Zmed - Zmin
    A2= Zmed - Zmax

    if (newwin<=Smax):

        #% Nivel A
        if (A1>0) & (A2<0):
            #%nivel B
            B1= float(Zxy)-float(Zmin)
            B2= float(Zxy)-float(Zmax)

            if (B1>0) & (B2<0):
                px=Zxy
            else:
                px=Zmed

        else:
            # Incrementamos el tamaño de
            # la ventana
            newwin= newwin+1
    else:
        px=Zxy

    return(px, newwin)



#aplicamos el filtro mediana_adaptiva
filtro3 = filtro_mediana_adaptiva(res3)

fig, ax = plt.subplots(1,2)
fig.set_size_inches(15,15)
ax[0].imshow(res3, cmap='gray')
ax[0].set_title(f"img con ruido impulsional salt")
ax[1].imshow(filtro3, cmap='gray')
ax[1].set_title(f"img filtrado mediana adaptativa")
plt.show()
```

###Filtro de media contra-armonica (filtro lieal)

Para la última imagen, que contiene ruido impulsional tipo pepper, se utilizó un filtro lineal llamado filtro de media contra-armónica. Este filtro fue elegido porque el ruido pepper introduce píxeles negros aleatorios (valor 0), y al utilizar un valor positivo de Q, el filtro minimiza el efecto de los valores bajos, eliminando eficazmente ese tipo de ruido. A diferencia de otros filtros, la media contra-armónica preserva mejor los detalles y bordes de la imagen, ya que no reemplaza los píxeles por valores extremos ni medianas, sino que ajusta su cálculo según la distribución local. Es especialmente útil cuando el ruido impulsivo afecta principalmente un extremo del rango de valores, como en este caso.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import scipy.ndimage as ndi

def filtro_contra_armonica(A,Q):

    try:
        #validamos que el denominador no sea cero
        den = (A**Q).sum()
    except:
        den = 0

    return (A**(Q+1)).sum()/den

filtro4 = ndi.generic_filter(res4,filtro_contra_armonica,[3,3],extra_keywords={'Q':2})

fig, ax = plt.subplots(1,2)
fig.set_size_inches(15,15)
ax[0].imshow(res4, cmap='gray')
ax[0].set_title(f"img con ruido impulsional pepper")
ax[1].imshow(filtro4, cmap='gray')
ax[1].set_title(f"img filtrado de media contra-armonica")
plt.show()
```

### 1.Restaure la imagen del siguiente código empleando el filtro de Wiener y el filtro Paramétrico. Determine el parámetro que mejor resultado genere (no modifique el código)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('cameraman.png', cv2.IMREAD_GRAYSCALE)
gray = cv2.normalize(img.astype('float'), None, 0.0, 1.0, cv2.NORM_MINMAX)
F = np.fft.fft2(gray)
vector = np.linspace(-0.5, 0.5, gray.shape[0])
U,V = np.meshgrid(vector, vector)
a = 5
b = 1
UV = U*a+V*b
G = F*np.fft.fftshift(np.sinc(np.pi*UV)*np.exp(-1j*np.pi*UV))
g = np.real(np.fft.ifft2(G))
plt.figure()
plt.imshow(g, cmap='gray')
plt.show()
```

De la documentación del curso disponible en GitHub se extrajeron los siguientes códigos, los cuales fueron utilizados y adaptados para trabajar con la imagen con ruido proporcionada en el enunciado de la tarea.

```python
#filtro Wiener
im1=g
import numpy.matlib
from matplotlib import cm
from matplotlib.pyplot import figure, draw, pause, imshow, show

H=np.fft.fftshift(np.sinc(np.pi*UV)*np.exp(-1j*np.pi*UV))

K = 0.0005
G = np.fft.fft2(im1)
W = np.conj(H) / (np.abs(H)**2 + K)
F_W = W * G
res_w = np.real(np.fft.ifft2(F_W))



fig, ax = plt.subplots(1,2)
fig.set_size_inches(15,15)
ax[0].imshow(g, cmap='gray')
ax[0].set_title(f"img original")
ax[1].imshow(res_w, cmap='gray')
ax[1].set_title(f"resultado img filtrado Wiener")
plt.show()
```

Para obtener la mejor imagen posible, se probaron diversos valores de K en el filtro Wiener. Se llegó a la conclusión de que con un valor alrededor de 0.0005 se lograba una buena reducción del ruido, preservando adecuadamente los contornos y detalles de la imagen. Con valores inferiores a este, aparecían sombras indeseadas que parecían múltiples versiones exageradas de los contornos. Por otro lado, con valores cercanos a 1, la imagen también se deterioraba, y aún se observaba un desenfoque marcado en el rostro. Por ello, un valor intermedio como 0.0005 resultó ser el más balanceado entre reducción de ruido y conservación de detalles.

```python
import cv2
import numpy as np
from matplotlib.pyplot import figure, draw, pause

im2=g
m, n = im2.shape
G = np.fft.fft2(im2)

p = np.array([[0, -1, 0],[-1, 4, -1],[0, -1, 0]])
Pp = np.fft.fft2(p, s=[m, n])

gamma = 0.01

F_param = (np.conj(H) * G) / (np.abs(H)**2 + gamma * np.abs(Pp)**2)
res_p = np.real(np.fft.ifft2(F_param))



fig, ax = plt.subplots(1,2)
fig.set_size_inches(15,15)
ax[0].imshow(g, cmap='gray')
ax[0].set_title(f"imagen original")
ax[1].imshow(res_p, cmap='gray')
ax[1].set_title(f"resultado img filtrado parametrico")
plt.show()
```

Utilizando el filtro paramétrico, también se realizaron múltiples pruebas para determinar el valor adecuado de gamma que permitiera eliminar el ruido de la imagen entregada. Se escogió el valor 0.05, ya que representaba un punto intermedio entre los distintos valores probados y proporcionaba el mejor equilibrio visual en la imagen restaurada. Con valores menores, aparecían múltiples versiones exageradas de los contornos, haciéndolos demasiado marcados. Por otro lado, con valores mayores, la imagen se veía excesivamente borrosa, perdiéndose los detalles importantes. Por esta razón, el valor 0.01 fue el más adecuado para obtener la nitidez y reducir el ruido simultáneamente.
