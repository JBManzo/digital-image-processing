###2. Genere una imagen ruido frecuencial de 10Hz, una segunda imagen con 30Hz y una tercera imagen con 50Hz en forma vertical respectivamente (usted define la amplitud de la señal).

El código utilizado para generar las imágenes con ruido senoidal fue extraído del material de apoyo entregado por el profesor, específicamente del archivo denominado "freq_noise" disponible en el repositorio de GitHub asociado al curso.
```python
import cv2
import numpy as np
import numpy.matlib
import matplotlib.pyplot as plt

img = img_2
print("Dimensiones img recortada (alto, ancho):", img.shape)
gray = cv2.normalize(img.astype('float'), None, 0.0, 1.0, cv2.NORM_MINMAX)

#ruido 10Hz
V = np.linspace(0,1,num=512)
Y = 0.2*np.sin(2*np.pi*10*V)
M = np.matlib.repmat(Y,512,1)
noise_img1= np.add(M, gray)

#ruido 30Hz
V = np.linspace(0,1,num=512)
Y = 0.2*np.sin(2*np.pi*30*V)
M = np.matlib.repmat(Y,512,1)
noise_img2= np.add(M, gray)

#ruido 50Hz
V = np.linspace(0,1,num=512)
Y = 0.2*np.sin(2*np.pi*50*V)
M = np.matlib.repmat(Y,512,1)
noise_img3= np.add(M, gray)

fig, ax = plt.subplots(1,3)
fig.set_size_inches(15,15)
ax[0].imshow(noise_img1, cmap='gray')
ax[0].set_title(f"Ruido 10 Hz")
ax[1].imshow(noise_img2, cmap='gray')
ax[1].set_title(f"Ruido 30 Hz")
ax[2].imshow(noise_img3, cmap='gray')
ax[2].set_title(f"Ruido 50 Hz")
plt.show()
```

###3. Genere una imagen que atenúe el ruido empleando filtros en la frecuencia para cada imagen del paso #2

Esta primera parte del filtro aplica una transformada de Fourier bidimensional (FFT) a las 3 imágenes con ruido del ítem anterior  (con frecuencias de 10 Hz, 30 Hz y 50 Hz).
Se utiliza la función "np.fft.fft2" y luego se centra el espectro de frecuencias con "np.fft.fftshift" para facilitar el análisis visual. Posteriormente se visualizan los espectros obtenidos de las 3 imágenes con el objetivo de resaltar los picos de frecuencia generados por el ruido.

```python
F = np.fft.fft2(noise_img1)
fshift1 = np.fft.fftshift(F)

F = np.fft.fft2(noise_img2)
fshift2 = np.fft.fftshift(F)

F = np.fft.fft2(noise_img3)
fshift3 = np.fft.fftshift(F)

fig, ax = plt.subplots(1,3)
fig.set_size_inches(15,15)
ax[0].imshow(np.log(np.abs(fshift1)+1), cmap='gray')
ax[0].set_title(f"espectro img ruido 10 Hz")
ax[1].imshow(np.log(np.abs(fshift2)+1), cmap='gray')
ax[1].set_title(f"espectro img ruido 30 Hz")
ax[2].imshow(np.log(np.abs(fshift3)+1), cmap='gray')
ax[2].set_title(f"Ruido img ruido 50 Hz")
plt.show()

frecuencias = [10, 30, 50]
fshifts = [fshift1, fshift2, fshift3]

# Ciclo para aplicar el mismo proceso de anulación de frecuencias a cada espectro
for i in range(3):
    frec = frecuencias[i]
    fshift = fshifts[i]

    centro_fila = 256
    centro_col = 256

    col_izq = 256 - frec
    col_der = 256 + frec

    fshift[256 - 10:256 + 10, col_izq - 2:col_izq + 2] = 0
    fshift[256 - 10:256 + 10, col_der - 2:col_der + 2] = 0
```

Se desarrolla este ciclo con el objetivo de anular los picos de frecuencia asociados al ruido senoidal vertical presente en las imágenes. Para cada caso, se calcula la posición horizontal donde aparecen dichos picos en el espectro de Fourier, y se bloquea una franja rectangular centrada en esas coordenadas. El procedimiento fue implementado siguiendo el método enseñado por el docente durante las clases, utilizando como guía el código referenciado anteriormente. Dado que las dimensiones de la imagen son conocidas (512×512 píxeles), se define el centro del espectro en la posición (256, 256). A partir de la frecuencia correspondiente a cada imagen, se calculan las columnas donde se ubican los picos de ruido, y se eliminan anulando un rango de píxeles alrededor de cada uno.

```python
img_filt1 = np.abs(np.fft.ifft2(np.fft.ifftshift(fshift1)))
img_filt2 = np.abs(np.fft.ifft2(np.fft.ifftshift(fshift2)))
img_filt3 = np.abs(np.fft.ifft2(np.fft.ifftshift(fshift3)))

fig, ax = plt.subplots(1,3)
fig.set_size_inches(15,15)
ax[0].imshow(img_filt1, cmap='gray')
ax[0].set_title(f"img ruido 10Hz filtrada")
ax[1].imshow(img_filt2, cmap='gray')
ax[1].set_title(f"img ruido 30Hz filtrada")
ax[2].imshow(img_filt3, cmap='gray')
ax[2].set_title(f"img ruido 50Hz filtrada")
plt.show()

fig, ax = plt.subplots(1,3)
fig.set_size_inches(15,15)
ax[0].imshow(noise_img1, cmap='gray')
ax[0].set_title(f"Ruido 10 Hz")
ax[1].imshow(noise_img2, cmap='gray')
ax[1].set_title(f"Ruido 30 Hz")
ax[2].imshow(noise_img3, cmap='gray')
ax[2].set_title(f"Ruido 50 Hz")
plt.show()
```

Por último, este bloque de código aplica la transformada inversa de Fourier (IFFT) a los espectros previamente filtrados para reconstruir las imágenes en el dominio espacial. Para ello, se utiliza la función "np.fft.ifftshift" que recentra el espectro antes de aplicar "np.fft.ifft2", permitiendo obtener correctamente la imagen reconstruida. Finalmente, se visualizan las tres imágenes resultantes con el ruido atenuado, junto con las imágenes originales con ruido en la parte inferior, con el objetivo de comparar y verificar visualmente la efectividad del filtrado en frecuencia aplicado a cada caso.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('cameraman.png',cv2.IMREAD_GRAYSCALE)
m = img.shape[0]
delta = 15
V = np.fix(np.linspace(delta,m-delta,delta)).astype('uint8')
img[V,:] = img[V,:]+50
img[:,V] = img[:,V]+50
plt.imshow(img, cmap='gray')
plt.show()

import cv2
import numpy as np
import numpy.matlib
import matplotlib.pyplot as plt
from matplotlib import cm

noise_img = cv2.normalize(img.astype('float32'), None, 0.0, 1.0, cv2.NORM_MINMAX)

F = np.fft.fft2(noise_img)
FS = np.fft.fftshift(F)


fc=256  # Frecuencia de corte
fa=50  # Frecuencia de corte
fb=15 # Frecuencia de corte


N= 4  # orden del filtro

x= np.linspace(-fc/2, fc/2, fc)
y= np.linspace(-fc/2, fc/2, fc)
X, Y = np.meshgrid(x,y)

H1 = 1/ (1+(np.power(np.sqrt( X**2+Y**2 )/fa,2*N)))
H2 = 1- 1/ (1+(np.power(np.sqrt( X**2+Y**2)/fb,2*N)))

H = 1- H1 * H2
H= H - np.min(H)
H= H / np.max(H)


BF = H*FS

spectrum = 0.1*np.log(1+np.abs(BF))
#cv2.imshow('fft2', spectrum)
#cv2.waitKey(0)

S= np.fft.ifft2(np.fft.fftshift(BF))
out = cv2.normalize(abs(S), None, 0.0, 1.0, cv2.NORM_MINMAX)



plt.imshow(noise_img, cmap='gray')
plt.title("Imagen con ruido (líneas)")
plt.show()

plt.imshow(out, cmap='gray')
plt.title("Imagen sin ruido")
plt.show()
```

Este código corresponde a una modificación del filtro parabanda Butterworth presentado en el repositorio del curso. Comienza normalizando la imagen ruidosa al rango [0, 1] para trabajar en punto flotante. Luego se aplica la transformada de Fourier bidimensional (FFT) a la imagen, y se centra el espectro usando "np.fft.fftshift". Posteriormente, se utiliza el filtro de rechazo de banda en el dominio de la frecuencia, se definen las 2 frecuencias de corte (fa y fb) y un orden N. Luego se construye el filtro el cual corresponde a una combinación de un filtro pasa bajo (H1) y un pasa alto (H2) de tipo Butterworth, de modo que el producto H1 * H2 genera una banda de paso que luego se invierte para formar una zona de rechazo.
