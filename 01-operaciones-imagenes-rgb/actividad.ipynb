###**1.Descargue una imagen del rostro de una persona (puede ser usted mismo) en color. La imagen debe tener aproximadamente 600x600 píxeles y debe poseer colores (no debe estar en escala de grises). Despliegue la imagen por el browser o una ventana.**

Para el primer ítem se escogió una imagen de Justin Bieber, un cantante famoso, la imagen tiene por dimensiones 1309 píxeles de alto, 736 de ancho y 3 canales.

Durante toda la tarea se utilizan las bibliotecas "OpenCV" la cual es utilizada para visualización y procesamiento de imágenes, "Matplotlib" la cual nos permite graficar y visualizar imágenes y datos y "NumPy" que nos permite realizar los cálculos matemáticos pertinentes.

Se utiliza la función "imread" para cargar la imagen e "imshow" para su visualización, en este caso se muestra la imagen tal cual es cargada (BGR) y con los canales ordenados (RGB).

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('justin.jpeg')

img2=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


plt.figure(figsize=(9, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title('BGR')


plt.subplot(1, 2, 2)
plt.imshow(img2)
plt.title('RGB')

plt.tight_layout()
plt.show()

print("Dimensiones (alto, ancho, canales):", img.shape)
```

###2.**Separe** los canales de la imagen RGB (paso 1) a tres variables con sus respectivos canales (R, G, y B). Luego almacene el resultado de dichas imágenes en diferentes archivos independientes (tip: investigue sobre el comando imwrite() en OpenCV)

Se separan los canales con la función "split()" y se almacenan en las variables "blue", "green" y "red" que es el orden en el que están desde un principio (BGR).

```python
blue, green, red = cv2.split(img)

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(blue, cmap='gray')
plt.title('Blue')


plt.subplot(1, 3, 2)
plt.imshow(green, cmap='gray')
plt.title('Green')

plt.subplot(1, 3, 3)
plt.imshow(red, cmap='gray')
plt.title('Red')

plt.tight_layout()
plt.show()
```

Luego con la función "imwrite" que nos permite almacenar la imagen en cualquier dispositivo y en cualquier formato, se guardan las imágenes en 3 archivos separados en los que su nombre indicará a qué canal pertenece cada uno.

###3.**Transforme** la imagen a niveles de gris empleando la función de transformación (cv2.COLOR_BGR2GRAY). Despliegue la imagen por el browser y compare esta imagen con cada uno de los canales R, G y B del paso 2. Para ello calcule la imagen diferencia. Despliege el resultado por el browser o una ventana.

Se utiliza la función "cvtColor" en la cual ingresando la imagen y la función de cambio que en este caso es de RGB a gris (cv2.COLOR_RGB2GRAY) nos da como resultado la imagen en escala de grises, posterior a esto se calcula la imagen diferencia de la resultante en niveles de gris con cada uno de los canales(R, G y B) usando la función "absdiff".

*Referencia: https://omes-va.com/adicion-sustraccion-de-imagenes/*

```python
img_gray=cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.imshow(img_gray, cmap='gray')
plt.title('niveles de gris')
plt.show()

dif_R=cv2.absdiff(img_gray, red)
dif_G=cv2.absdiff(img_gray, green)
dif_B=cv2.absdiff(img_gray, blue)

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(dif_R,cmap='gray')
plt.title('img_gray - red')


plt.subplot(1, 3, 2)
plt.imshow(dif_G,cmap='gray')
plt.title('img_gray - green')

plt.subplot(1, 3, 3)
plt.imshow(dif_B,cmap='gray')
plt.title('img_gray - blue')

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(red,cmap='gray')
plt.title('Red')


plt.subplot(1, 3, 2)
plt.imshow(green,cmap='gray')
plt.title('Green')

plt.subplot(1, 3, 3)
plt.imshow(blue,cmap='gray')
plt.title('Blue')

plt.tight_layout()
plt.show()
```

###4.**Ecualice** los histogramas de los canales R, G y B, obtenidos en el paso 2 según el algoritmo visto en clases y muestre el resultado por el browser con el resultado de cada ecualización.

Se utiliza la función "equalizedHist" vista en clases para cada uno de los canales para compararlos con los canales sin ecualizar.

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('justin.jpeg')

blue, green, red =  cv2.split(img)

eq_blue = cv2.equalizeHist(blue)
eq_green = cv2.equalizeHist(green)
eq_red = cv2.equalizeHist(red)


plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(eq_blue,cmap='gray')
plt.title('blue ecualizado')


plt.subplot(1, 3, 2)
plt.imshow(eq_green,cmap='gray')
plt.title('green ecualizado')

plt.subplot(1, 3, 3)
plt.imshow(eq_red,cmap='gray')
plt.title('red ecualizado')

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(blue,cmap='gray')
plt.title('blue')


plt.subplot(1, 3, 2)
plt.imshow(green ,cmap='gray')
plt.title('green')

plt.subplot(1, 3, 3)
plt.imshow(red,cmap='gray')
plt.title('red')

plt.tight_layout()
plt.show()
```

###5.**Modifique** las imágenes de paso 3 del siguiente modo. Utilice la función gamma con valor 0.5 sobre el canal R, 2.5 sobre el canal G y 0.1 sobre el canal B. Almacene estos resultados en archivos independientes y despliegue el resultado por el browser.

Para realizar la corrección gamma se necesita definir la función "gamma_correction"(según el material del curso, ppt3), la cual normaliza cada pixel en el rango 0 y 1 para posteriormente aplicar la corrección gamma y convertir la imagen de nuevo a su rango original (0 y 255) en tipo unit8.

```python
#imagenes del paso 3: dif_R, dif_G, dif_B
import cv2
import numpy as np
import matplotlib.pyplot as plt

auxB = dif_B
auxG = dif_G
auxR = dif_R

def gamma_correction(img, factor):
  img = img/255.0
  img = cv2.pow(img, factor)
  return np.uint8(img*255)



gammaB = gamma_correction(auxB, 0.1)
gammaG = gamma_correction(auxG, 2.5)
gammaR= gamma_correction(auxR, 0.5)

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.imshow(gammaB ,cmap='gray')
plt.title('gamma B')


plt.subplot(1, 3, 2)
plt.imshow(gammaG ,cmap='gray')
plt.title('gamma G')

plt.subplot(1, 3, 3)
plt.imshow(gammaR ,cmap='gray')
plt.title('gamma R')

plt.tight_layout()
plt.show()

filename4 = 'justinGammaB.jpg'
filename5 = 'justinGammaG.jpg'
filename6 = 'justinGammaR.jpg'

cv2.imwrite(filename4, gammaB)
cv2.imwrite(filename5, gammaG)
cv2.imwrite(filename6, gammaR)
```
